C.) Two major challenges that make Natural Language Processing (NLP) complex are:
Ambiguity and Context Understanding
Language is inherently ambiguous because words and sentences can have multiple meanings depending on the context. For example, the word "bank" could mean a financial institution or the side of a river. NLP systems need to accurately interpret the correct meaning based on context, which requires sophisticated models for semantic, syntactic, and pragmatic analysis. Achieving this level of understanding remains a significant challenge due to the variety of linguistic structures and contexts.

Multilingualism and Language Diversity
Human languages are highly diverse in syntax, grammar, vocabulary, and cultural nuances. Building NLP models that effectively handle multiple languages—especially low-resource languages with limited data—is difficult. Cross-lingual transfer and multilingual understanding require advanced techniques and extensive training data to ensure NLP systems work well globally across different languages

```python
#Extract the following patterns using regex:
# Using re.findall to extract all email addresses
import re

# Sample text
text = "Contact us at support@company.com or sales@business.org. For more, email info@service.net."

# Extract email addresses
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)
# \b: Matches word boundaries to ensure the email is a standalone word
# [\w.-]+: Matches one or more word characters (letters, digits, underscores) or the characters '.' and '-'
# @: Matches the literal '@' symbol
# [\w.-]+: Matches the domain name (e.g., example, website)
# \.\w+: Matches a '.' followed by one or more word characters (e.g., com, org)
# \b: Ensures the match ends at a word boundary

print("Emails Found:", emails)  # Prints the list of extracted email addresses
```

```python
import string

text = "NLP makes AI smarter! But, sometimes, it’s challenging... Don’t you agree?"

# 1. Remove punctuation
clean_text = text.translate(str.maketrans("", "", string.punctuation))

# 2. Convert to lowercase
clean_text = clean_text.lower()

# 3. Split into words
words = clean_text.split()

# 4. Keep only alphabetic tokens (remove any stray non-alphabetic)
clean_words = [w for w in words if w.isalpha()]

print("Original text:", text)
print("Cleaned text:", clean_text)
print("Words:", words)
print("Cleaned Tokens:", clean_words)
```

```python
# text preprocessing
import re
import string

text = "OMG!! NLP is soooo coool 🤩...!!! It costs $1000. Learn it now at https://3mtt.com 😎."

# Remove emojis and non-ASCII characters
text = text.encode('ascii', 'ignore').decode('ascii')

# Remove URLs
text = re.sub(r'http\S+', '', text)

# Remove numbers
text = re.sub(r'\d+', '', text)

# Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

# Convert to lowercase
text = text.lower()

# Split into words
words = text.split()

print(words)
```

```python
#Python code to generate one-hot encoded vectors
# Define vocabulary

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Sample data: each word is its own row
data = pd.DataFrame({
    "vocabulary": ["apple", "banana", "milk", "perfume", "car"]
})

# One-Hot Encoding
encoder = OneHotEncoder(sparse_output=False)
one_hot = encoder.fit_transform(data[["vocabulary"]])

# Put result into DataFrame
one_hot_df = pd.DataFrame(one_hot, columns=encoder.get_feature_names_out(["vocabulary"]))

# Combine with original data for clarity
final_df = pd.concat([data, one_hot_df], axis=1)

print(final_df)
```

```python
# BOW
from sklearn.feature_extraction.text import CountVectorizer

# Example Sentences
documents = ["The quick brown fox jumps over the lazy dog.", "The dog sleeps in the kernel"]

# Create BoW Model
vectorizer = CountVectorizer()
bow = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("Bag of Words Representation:\n", bow.toarray())
```

```python
# TfidfVectorizer.
from sklearn.feature_extraction.text import TfidfVectorizer

# Example Sentences
documents = ["The quick brown fox jumps over the lazy dog.", "The dog sleeps in the kernel"]

# Create TF-IDF Model
tfidf_vectorizer = TfidfVectorizer()
tfidf = tfidf_vectorizer.fit_transform(documents)

print("Vocabulary:", tfidf_vectorizer.get_feature_names_out())
print("TF-IDF Representation:\n", tfidf.toarray())
```

```python
# WORD2VEC
from gensim.models import Word2Vec

# Example Sentences
sentences = [["the", "dog", "barks"], ["the", "cat", "meows"], ["the", "bird", "sings."]]

# Train Word2Vec Model
word2vec_model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, workers=4)

# Get Embedding for a Word
print("Word Embedding for 'dog':\n", word2vec_model.wv['dog'])
```

```python
#PYTHON WORD EMBEDDINGS

from gensim.models import Word2Vec

# Example text (split into tokenized sentences)
sentences = [
    ["the", "cat", "meows"],
    ["the", "dog", "barks"],
    ["the", "bird", "sings"]
]

# 1. Train Word2Vec model
word2vec_model = Word2Vec(
    sentences,
    vector_size=10,   # size of word embeddings
    window=2,         # context window size
    min_count=1,      # ignore words with total frequency lower than this
    workers=4,        # use 4 CPU cores
    sg=0              # 0 = CBOW, 1 = Skip-gram
)

# 2. Retrieve embedding for the word "dog"
dog_vector = word2vec_model.wv["dog"]

print("Word Embedding for 'dog':\n", dog_vector)
```
